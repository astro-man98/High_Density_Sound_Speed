```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
RVSS_List = {"V" : 'red',  
             "IV" : 'purple',
             "III" : 'blue',
             "II": 'green',
             "I": 'black'}  
```

```python
A_1p4_data = np.loadtxt('./Figure_Data/Max_cs/EOS_V_1p4.txt')
A_1p3_data = np.loadtxt('./Figure_Data/Max_cs/EOS_V_1p3.txt')
B_1p4_data = np.loadtxt('./Figure_Data/Max_cs/EOS_III_1p4.txt')
B_1p3_data = np.loadtxt('./Figure_Data/Max_cs/EOS_III_1p3.txt')
E_1p3_data = np.loadtxt('./Figure_Data/Max_cs/EOS_IV_1p3.txt')
E_1p35_data = np.loadtxt('./Figure_Data/Max_cs/EOS_IV_1p35.txt')
E_1p4_data = np.loadtxt('./Figure_Data/Max_cs/EOS_IV_1p4.txt')
C_1p3_data = np.loadtxt('./Figure_Data/Max_cs/EOS_II_1p3.txt')
```

```python
axes_label_fontsize = 18
tick_label_fontsize = 12
text_fontsize = 16

max_cs2_fig,max_cs2_axes = plt.subplots(1)

max_cs2_axes.plot(A_1p4_data[:,0]*1/203,A_1p4_data[:,1],color = RVSS_List["V"],label='EOS-V 1.4M - 1.4M')
max_cs2_axes.plot(A_1p3_data[:,0]*1/203,A_1p3_data[:,1],color = RVSS_List["V"],label='EOS-V 1.3M - 1.3M',linestyle='--')
max_cs2_axes.plot(E_1p4_data[:,0]*1/203,E_1p4_data[:,1],color = RVSS_List["IV"],label='EOS-IV 1.4M - 1.4M')
max_cs2_axes.plot(E_1p35_data[:,0]*1/203,E_1p35_data[:,1],color = RVSS_List["IV"],label='EOS-IV 1.35M - 1.35M',linestyle='-.')
max_cs2_axes.plot(E_1p3_data[:,0]*1/203,E_1p3_data[:,1],color = RVSS_List["IV"],label='EOS-IV 1.3M - 1.3M',linestyle='--')
max_cs2_axes.plot(B_1p4_data[:,0]*1/203,B_1p4_data[:,1],color = RVSS_List["III"],label='EOS-III 1.4M - 1.4M')
max_cs2_axes.plot(B_1p3_data[:,0]*1/203,B_1p3_data[:,1],color = RVSS_List["III"],label='EOS-III 1.3M - 1.3M',linestyle='--')
max_cs2_axes.plot(C_1p3_data[:,0]*1/203,C_1p3_data[:,1],color = RVSS_List["II"],label='EOS-II 1.3M - 1.3M',linestyle='--')
max_cs2_axes.set_xlabel('t (ms)',fontsize=axes_label_fontsize)
max_cs2_axes.set_ylabel(r'$(c_s^2)_{\text{max}}$',fontsize=axes_label_fontsize)
max_cs2_axes.tick_params(labelsize=tick_label_fontsize)
max_cs2_axes.legend()
max_cs2_fig.tight_layout()

max_cs2_fig.savefig('./Figures/Final_Paper_Figs/BNS_maxcs2.pdf',format = 'pdf',dpi = 300)
```
